```python
import pandas as pd
import numpy as np

pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.large_repr", "truncate")
```

```python
#Data Inspection
original_data = pd.read_csv("../data/Crime_Data_from_2010_to_2019.csv")
original_data.head()
```

```python
data = original_data.copy()
```

```python
# Columns to drop - they are not important
cols_to_drop = [
    "AREA ",        
    "Rpt Dist No",
    "Mocodes",
    "Crm Cd 1",
    "Crm Cd 2",
    "Crm Cd 3",
    "Crm Cd 4",
    "Cross Street",
    "LOCATION"
]

# Drop columns
data = data.drop(columns=[col for col in cols_to_drop if col in data.columns])

# Check the result
data.columns
```

```python
data = data.rename(columns={"DR_NO": "Id"})
data = data.rename(columns={"DATE OCC": "Date Occ"})
data = data.rename(columns={"TIME OCC": "Time Occ"})
data.columns
```

```python
# Convert strings to datetime
data["Date Rptd"] = pd.to_datetime(data["Date Rptd"],format="%Y %b %d %I:%M:%S %p")
data["Date Occ"] = pd.to_datetime(data["Date Occ"], format="%Y %b %d %I:%M:%S %p")
# Format as dd-mm-yyyy
data["Date Rptd"] = data["Date Rptd"].dt.strftime("%d-%m-%Y")
data["Date Occ"] = data["Date Occ"].dt.strftime("%d-%m-%Y")

data.head()
```

```python
# Ensure time is string and pad with zeros if needed
data['Time Occ'] = data['Time Occ'].astype(str).str.zfill(4)
# Convert to proper HH:MM format
data['Time Occ'] = data['Time Occ'].str[:2] + ':' + data['Time Occ'].str[2:]

data.head()
```

```python
# Replace invalid ages with Unknown
data["Vict Age"] = data["Vict Age"].apply(lambda x: x if 0 < x <= 100 else 'Unknown')
print(data["Vict Age"].unique())
```

```python
# Replace invalid sex values with X
data["Vict Sex"] = data["Vict Sex"].apply(
    lambda x: x if x in ["M", "F"] else "X"
)
print(data['Vict Sex'].unique())
```

```python
category_rules = {

    # ---------------- RESIDENTIAL ----------------
    "Residential Interior": [
        "DWELLING", "APARTMENT", "RESIDENCE",
        "CONDOMINIUM", "HOUSING", "GROUP HOME",
        "MOBILE HOME", "SRO", "DORMITORY",
        "FRAT", "SORORITY", "FOSTER HOME",
        "VACATION RENTAL", "SHELTER"
    ],

    "Residential Exterior": [
        "PORCH", "GARAGE", "CARPORT",
        "BALCONY", "DRIVEWAY", "YARD"
    ],

    # ----------------  PARKING AREA ----------------
    "Parking Area": [
        "PARKING", "VALET"
    ],

    # ---------------- PUBLIC SPACE ----------------
    "Street/Public Outdoor": [
        "STREET", "SIDEWALK", "ALLEY",
        "FREEWAY", "VACANT",
        "BEACH", "RIVER", "BRIDGE",
        "UNDERPASS", "PEDESTRIAN",
        "OUTSIDE", "ENCAMPMENT",
        "PARK", "PLAYGROUND"
    ],

    # ---------------- VEHICLE  ---------------- 
    "Vehicle": [
        "VEHICLE", "AUTO", "TRUCK"
    ],

    # ---------------- STORE / RETAIL ----------------
"Store/Retail": [
    "STORE", "MARKET", "SHOP", "DEPARTMENT", 
    "MINI-MART", "GROCERY", "OTHER STORE"
],

# ---------------- COMMERCIAL / SERVICE ----------------
"Commercial/Business": [
    "RESTAURANT", "SALON", "REPAIR",
    "HOTEL", "MOTEL",
    "WAREHOUSE", "STORAGE",
    "FACTORY", "MANUFACTURING",
    "OFFICE", "COMPANY", "CONVENTION",
    "CHECK CASHING", "FINANCE", "CREDIT UNION",
    "DIY CENTER", "LAUNDROMAT",
    "CLEANER", "TATTOO", "MASSAGE", "CAR WASH"
],

    # ---------------- EDUCATION ----------------
    "Educational": [
        "SCHOOL", "COLLEGE",
        "UNIVERSITY", "TRADE SCHOOL"
    ],

    # ---------------- RECREATION ----------------
    "Entertainment/Recreation": [
        "THEATRE", "CLUB", "BAR",
        "ARENA", "SPORT",
        "BOWLING", "POOL",
        "AMUSEMENT", "STADIUM",
        "CENTER", "ARCADE",
        "GAME ROOM", "SKATING",
        "COURTS", "MASS GATHERING",
        "SPA", "GYM",
        "GOLF COURSE", "MARINA", "DOCK"
    ],

    # ---------------- TRANSPORT INFRASTRUCTURE ----------------
    "Transportation/Transit": [
        "MTA", "BUS", "TRAIN",
        "SUBWAY", "LINE",
        "STATION", "TRANSIT",
        "TAXI", "AIRCRAFT",
        "AIRPORT", "TRACK",
        "TUNNEL"
    ],

    # ---------------- GOVERNMENT ----------------
    "Government/Institutional": [
        "POLICE", "FIRE STATION",
        "GOVERNMENT", "POST OFFICE",
        "DETENTION", "JAIL",
        "DEFENSE"
    ],

    # ---------------- HEALTH ----------------
    "Healthcare/Medical": [
        "HOSPITAL", "CLINIC",
        "MEDICAL", "NURSING",
        "CARE", "HOSPICE",
        "METHADONE"
    ],

    # ---------------- CULTURAL ----------------
    "Religious/Cultural/Civic": [
        "CHURCH", "MOSQUE",
        "SYNAGOGUE", "TEMPLE",
        "LIBRARY", "MUSEUM",
        "WORSHIP", "CULTURAL",
        "MONUMENT", "CEMETARY",
        "MORTUARY"
    ]
}
```

```python
def categorize_premis(desc):
    if pd.isna(desc):
        return "Unknown"

    desc = desc.upper()

    for category, keywords in category_rules.items():
        if any(keyword in desc for keyword in keywords):
            return category

    return "Other"


data["Premis Category"] = data["Premis Desc"].apply(categorize_premis)
```

```python
premis_mapping_check = (
    data
    .groupby(["Premis Desc", "Premis Category"])
    .size()
    .reset_index(name="Count")
    .sort_values("Count", ascending=False)
)

premis_mapping_check
```
