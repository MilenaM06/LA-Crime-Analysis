```python
import pandas as pd
import numpy as np

pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.large_repr", "truncate")
```

```python
#Data Inspection
original_data = pd.read_csv("../data/Crime_Data_from_2010_to_2019.csv")
original_data.head()
```

```python
original_data.shape
```

```python
original_data.info()
```

```python
data = original_data.copy()
```

```python
# Columns to drop - they are not important
cols_to_drop = [
    "AREA ",        
    "Rpt Dist No",
    "Part 1-2",
    "Crm Cd",
    "Mocodes",
    "Premis Cd",
    "Weapon Used Cd",
    "Status",
    "Status Desc",
    "Crm Cd 1",
    "Crm Cd 2",
    "Crm Cd 3",
    "Crm Cd 4",
    "LOCATION",
    "Cross Street",

]

# Drop columns
data = data.drop(columns=[col for col in cols_to_drop if col in data.columns])

# Check the result
data.columns
```

```python
data = data.rename(columns={"DR_NO": "ID"})
data = data.rename(columns={"Date Rptd": "DATE RPTD"})
data = data.rename(columns={"Crm Cd Desc": "CRIME"})
data = data.rename(columns={"Vict Age": "VICT AGE"})
data = data.rename(columns={"Vict Sex": "VICT SEX"})
data = data.rename(columns={"Vict Descent": "VICT DESCENT"})
data = data.rename(columns={"Premis Desc": "PLACE"})
data = data.rename(columns={"Weapon Desc": "WEAPON"})
data.columns
```

```python
# Convert strings to datetime
data["DATE RPTD"] = pd.to_datetime(data["DATE RPTD"],format="%Y %b %d %I:%M:%S %p")
data["DATE OCC"] = pd.to_datetime(data["DATE OCC"], format="%Y %b %d %I:%M:%S %p")
# Format as dd-mm-yyyy
data["DATE RPTD"] = data["DATE RPTD"].dt.strftime("%d-%m-%Y")
data["DATE OCC"] = data["DATE OCC"].dt.strftime("%d-%m-%Y")

data.head()
```

```python
# Ensure time is string and pad with zeros if needed
data['TIME OCC'] = data['TIME OCC'].astype(str).str.zfill(4)
# Convert to proper HH:MM format
data['TIME OCC'] = data['TIME OCC'].str[:2] + ':' + data['TIME OCC'].str[2:]

data.head()
```

```python
data.isnull().sum()
```

```python
print(data['VICT AGE'].unique())
```

```python
# Replace invalid ages with Unknown
data["VICT AGE"] = data["VICT AGE"].apply(lambda x: x if 0 < x <= 100 else 'Unknown')
print(data["VICT AGE"].unique())
```

```python
print(data['VICT SEX'].unique())
```

```python
# Replace invalid sex values with X
data["VICT SEX"] = data["VICT SEX"].apply(
    lambda x: x if x in ["M", "F"] else "X"
)
print(data['VICT SEX'].unique())
```

```python
print(data["VICT DESCENT"].unique())
```

```python
valid_descent = ['A','B','C','D','F','G','H','I','J','K','L','O','P','S','U','V','W','X','Z']
# Replace invalid descent values with X
data["VICT DESCENT"] = data["VICT DESCENT"].apply(lambda x: x if x in valid_descent else "X")
print(data["VICT DESCENT"].unique())
```

```python
weapon_counts = data["WEAPON"].value_counts(dropna=False)
print(f"Number of rows: {weapon_counts.shape[0]}")
weapon_counts.to_frame()
```

```python
# Fill NaN with "No Weapon"
data["WEAPON"] = data["WEAPON"].fillna("No Weapon")

# Define 10 categories
weapon_mapping = {
    # Firearms
    "HAND GUN": "Firearm",
    "SEMI-AUTOMATIC PISTOL": "Firearm",
    "REVOLVER": "Firearm",
    "AIR PISTOL/REVOLVER/RIFLE/BB GUN": "Firearm",
    "RIFLE": "Firearm",
    "SHOTGUN": "Firearm",
    "AUTOMATIC WEAPON/SUB-MACHINE GUN": "Firearm",
    "ASSAULT WEAPON/UZI/AK47/ETC": "Firearm",
    "UZI SEMIAUTOMATIC ASSAULT RIFLE": "Firearm",
    "MAC-10 SEMIAUTOMATIC ASSAULT WEAPON": "Firearm",
    "MAC-11 SEMIAUTOMATIC ASSAULT WEAPON": "Firearm",
    "SEMI-AUTOMATIC RIFLE": "Firearm",
    "UNK TYPE SEMIAUTOMATIC ASSAULT RIFLE": "Firearm",
    "M-14 SEMIAUTOMATIC ASSAULT RIFLE": "Firearm",
    "M1-1 SEMIAUTOMATIC ASSAULT RIFLE": "Firearm",
    "RELIC FIREARM": "Firearm",
    "OTHER FIREARM": "Firearm",
    "UNKNOWN FIREARM": "Firearm",
    "SIMULATED GUN": "Firearm",
    "HECKLER & KOCH 93 SEMIAUTOMATIC ASSAULT RIFLE": "Firearm",
    "SAWED OFF RIFLE/SHOTGUN": "Firearm",
    "HECKLER & KOCH 91 SEMIAUTOMATIC ASSAULT RIFLE": "Firearm",
    "STARTER PISTOL/REVOLVER": "Firearm",
    "ANTIQUE FIREARM": "Firearm",

    # Knives / Sharp objects
    "KNIFE WITH BLADE 6INCHES OR LESS": "Knife/Sharp Object",
    "OTHER KNIFE": "Knife/Sharp Object",
    "FOLDING KNIFE": "Knife/Sharp Object",
    "KNIFE WITH BLADE OVER 6 INCHES IN LENGTH": "Knife/Sharp Object",
    "KITCHEN KNIFE": "Knife/Sharp Object",
    "DIRK/DAGGER": "Knife/Sharp Object",
    "BOWIE KNIFE": "Knife/Sharp Object",
    "CLEAVER": "Knife/Sharp Object",
    "STRAIGHT RAZOR": "Knife/Sharp Object",
    "RAZOR": "Knife/Sharp Object",
    "RAZOR BLADE": "Knife/Sharp Object",
    "UNKNOWN TYPE CUTTING INSTRUMENT": "Knife/Sharp Object",
    "OTHER CUTTING INSTRUMENT": "Knife/Sharp Object",
    "SWITCH BLADE": "Knife/Sharp Object",
    "AXE": "Knife/Sharp Object",
    "SCISSORS": "Knife/Sharp Object",
    "ICE PICK": "Knife/Sharp Object",
    "MACHETE": "Knife/Sharp Object",
    "SWORD": "Knife/Sharp Object",
    "SCREWDRIVER": "Knife/Sharp Object",
    "GLASS": "Knife/Sharp Object",
    
    # Strong-Arm (personal bodily force)
    "STRONG-ARM (HANDS, FIST, FEET OR BODILY FORCE)": "Strong-Arm",

    # Blunt objects
    "STICK": "Blunt Object",
    "CLUB/BAT": "Blunt Object",
    "BLUNT INSTRUMENT": "Blunt Object",
    "PIPE/METAL PIPE": "Blunt Object",
    "TIRE IRON": "Blunt Object",
    "HAMMER": "Blunt Object",
    "BOARD": "Blunt Object",
    "BELT FLAILING INSTRUMENT/CHAIN": "Blunt Object",
    "BLACKJACK": "Blunt Object",
    "FIXED OBJECT": "Blunt Object",
    "BOTTLE": "Blunt Object",
    "BRASS KNUCKLES": "Blunt Object",
    "ROCK/THROWN OBJECT": "Blunt Object",
    "CONCRETE BLOCK/BRICK": "Blunt Object",

    # Chemical
    "MACE/PEPPER SPRAY": "Chemical",
    "CAUSTIC CHEMICAL/POISON": "Chemical",
    "SCALDING LIQUID": "Chemical",
    "LIQUOR/DRUGS": "Chemical",

    # Explosive/Fire
    "EXPLOXIVE DEVICE": "Explosive/Fire",
    "BOMB THREAT": "Explosive/Fire",
    "FIRE": "Explosive/Fire",

    # Vehicle
    "VEHICLE": "Vehicle",

    # Threats
    "VERBAL THREAT": "Threat",
    "PHYSICAL PRESENCE": "Threat",
    "DEMAND NOTE": "Threat",

    # No Weapon
    "No Weapon": "No Weapon"
}
# Apply mapping
data["WEAPON CATEGORY"] = data["WEAPON"].map(weapon_mapping).fillna("Other")

# Drop original WEAPON column
#data.drop(columns=["WEAPON"], inplace=True)

# Check result
weapon_counts = data["WEAPON CATEGORY"].value_counts(dropna=False)
print(data.columns)
print(f"Number of rows: {weapon_counts.shape[0]}")
weapon_counts.to_frame()
```
