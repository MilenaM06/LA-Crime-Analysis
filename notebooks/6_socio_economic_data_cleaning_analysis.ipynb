```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import seaborn as sns
import geopandas as gpd
import numpy as np
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.large_repr", "truncate")
```

```python
#Data Inspection
district_data = pd.read_csv("../raw_data/socio_economic_data_2010.csv")
district_data
```

```python
data = district_data.copy()
data.head()
```

```python
# Assuming your DataFrame is called df
data['Council District'] = data['Council District'].str.split(' - ').str[0]

# Convert to integer if you want
data['Council District'] = data['Council District'].astype(int)

data
```

```python
data.info()
```

```python
# Convert numeric columns to float
numeric_cols = data.columns[1:]  # all except district number
for col in numeric_cols:
    data[col] = data[col].astype(str).str.replace(',', '').astype(float)

data
```

```python
districts_path = "../raw_data/LA_City_Council_Districts/LA_City_Council_Districts.shp"
districts_gdf = gpd.read_file(districts_path)
districts_gdf.head()
```

```python
# Ensure your districts_gdf has 'DISTRICT' column with district numbers
districts_gdf['DISTRICT'] = districts_gdf['DISTRICT']

# Create a color map with as many colors as districts
num_districts = len(districts_gdf)
colors = cm.tab20(np.linspace(0, 1, num_districts))

# Plot map
fig, ax = plt.subplots(1, 1, figsize=(12, 12))
districts_gdf.plot(ax=ax, color=colors, edgecolor='black')

# Add district numbers at the centroid of each district
for idx, row in districts_gdf.iterrows():
    centroid = row['geometry'].centroid
    ax.text(centroid.x, centroid.y, row['DISTRICT'], horizontalalignment='center',
            verticalalignment='center', fontsize=10, fontweight='bold', color='black')

ax.set_title('LA City Council Districts', fontsize=16)
ax.axis('off')
plt.show()
```

```python
# Sort by population
pop_df = data[['COUNCIL DISTRICT NUMBER', 'POPULATION']] \
    .sort_values('POPULATION', ascending=False)

# Extract correct order
district_order = pop_df['COUNCIL DISTRICT NUMBER']

# Bar plot
plt.figure(figsize=(12,6))
bars = sns.barplot(
    x='COUNCIL DISTRICT NUMBER',
    y='POPULATION',
    data=pop_df,
    order=district_order
)

# Add population values
for bar in bars.patches:
    height = bar.get_height()
    bars.annotate(
        f'{int(height):,}',
        xy=(bar.get_x() + bar.get_width()/2, height),
        xytext=(0,5),
        textcoords='offset points',
        ha='center',
        va='bottom',
        fontsize=10
    )

plt.title('Population by LA City Council District', fontsize=16)
plt.xlabel('Council District Number')
plt.ylabel('Population')

plt.show()
```

```python
# Select race columns
race_cols = [
    'WHITE', 'BLACK', 'AMERICAN INDIAN/ALASKA NATIVE', 'ASIAN',
    'NATIVE HAWAIIAN/PACIFIC ISLANDER', 'HISPANIC OR LATINO',
    'OTHER RACE', 'MULTI-RACIAL'
]

# Prepare dataframe
race_df = data[['COUNCIL DISTRICT NUMBER'] + race_cols] \
    .sort_values('COUNCIL DISTRICT NUMBER')

race_df.set_index('COUNCIL DISTRICT NUMBER', inplace=True)

# Convert to percentages
race_percent = race_df.div(race_df.sum(axis=1), axis=0) * 100

# Plot stacked percentage bar chart
race_percent.plot(
    kind='bar',
    stacked=True,
    figsize=(14,7),
    colormap='tab20'
)

plt.title('Racial Composition (%) by LA City Council District', fontsize=16)
plt.xlabel('Council District Number')
plt.ylabel('Population Share (%)')

plt.legend(title='Race', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
# Prepare data
sex_df = data[['COUNCIL DISTRICT NUMBER', 'MALE', 'FEMALE']].copy()

# Calculate total population per district
sex_df['TOTAL'] = sex_df['MALE'] + sex_df['FEMALE']

# Convert to percentages
sex_df['MALE_PCT'] = (sex_df['MALE'] / sex_df['TOTAL']) * 100
sex_df['FEMALE_PCT'] = (sex_df['FEMALE'] / sex_df['TOTAL']) * 100

# Plot grouped bar chart
fig, ax = plt.subplots(figsize=(14, 7))

width = 0.35
x = sex_df['COUNCIL DISTRICT NUMBER']

ax.bar(x - width/2, sex_df['MALE_PCT'], width, label='Male', color='#1f77b4')
ax.bar(x + width/2, sex_df['FEMALE_PCT'], width, label='Female', color='#ff7f0e')

# Add percentage labels
for i in range(len(x)):
    ax.text(
        x.iloc[i] - width/2,
        sex_df['MALE_PCT'].iloc[i] + 0.3,
        f"{sex_df['MALE_PCT'].iloc[i]:.1f}%",
        ha='center', va='bottom', fontsize=9
    )
    
    ax.text(
        x.iloc[i] + width/2,
        sex_df['FEMALE_PCT'].iloc[i] + 0.3,
        f"{sex_df['FEMALE_PCT'].iloc[i]:.1f}%",
        ha='center', va='bottom', fontsize=9
    )

ax.set_xlabel('Council District Number')
ax.set_ylabel('Population Share (%)')
ax.set_title('Population by Sex (%) in LA City Council Districts', fontsize=16)

ax.set_xticks(x)
ax.set_ylim(0, 60)   # percentage scale
ax.legend()

plt.tight_layout()
plt.show()
```

```python
### ---------- CHECK IF THIS CAN BE MORE INFORMATIVE ----------------###
# Columns for age groups
age_cols = [
    'AGE UNDER 5','AGE 5–9','AGE 10–14','AGE 15–19','AGE 20–24',
    'AGE 25–34','AGE 35–44','AGE 45–54','AGE 55–64','AGE 65–74',
    'AGE 75–84','AGE 85 AND UP'
]

# Prepare age data
age_df = data[['COUNCIL DISTRICT NUMBER'] + age_cols].copy()
age_df[age_cols] = age_df[age_cols].round().astype(int)

# Plot stacked bar chart
fig, ax = plt.subplots(figsize=(16, 8))

bottom = [0] * len(age_df)
colors = plt.cm.tab20.colors  # 20 distinct colors

for i, col in enumerate(age_cols):
    ax.bar(age_df['COUNCIL DISTRICT NUMBER'], age_df[col], bottom=bottom, label=col, color=colors[i % len(colors)])
    bottom = [bottom[j] + age_df[col][j] for j in range(len(bottom))]

ax.set_xlabel('Council District Number')
ax.set_ylabel('Population')
ax.set_title('Population by Age Groups in LA City Council Districts', fontsize=16)
ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')  # legend outside chart
plt.xticks(age_df['COUNCIL DISTRICT NUMBER'])
plt.tight_layout()
plt.show()
```

```python
# Sort by median age
data_sorted = data.sort_values('MEDIAN AGE', ascending=False)  # descending → oldest first

# Plot vertical bars
plt.figure(figsize=(14,6))
bars = plt.bar(
    x=data_sorted['COUNCIL DISTRICT NUMBER'].astype(str),  # treat as categorical
    height=data_sorted['MEDIAN AGE'],
    color='purple'
)

# Add values on top of bars
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,  # center of the bar
        height + 0.2,                     # slightly above bar
        f'{height:.1f}',
        ha='center',
        va='bottom',
        fontsize=9
    )

plt.xlabel('Council District Number')
plt.ylabel('Median Age')
plt.title('Median Age by LA City Council District (Sorted by Age)', fontsize=16)
plt.xticks(rotation=0)
plt.ylim(0, data_sorted['MEDIAN AGE'].max() + 5)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Sort data by average household size
data_sorted = data.sort_values('AVERAGE HOUSEHOLD SIZE', ascending=False)

plt.figure(figsize=(12,6))
bars = plt.bar(range(len(data_sorted)), data_sorted['AVERAGE HOUSEHOLD SIZE'], color='mediumpurple', alpha=0.8)

# Add numbers on top of bars
for i, height in enumerate(data_sorted['AVERAGE HOUSEHOLD SIZE']):
    plt.text(i, height + 0.02, f"{height:.2f}", ha='center', fontsize=9)

# Replace x-ticks with council district numbers
plt.xticks(range(len(data_sorted)), data_sorted['COUNCIL DISTRICT NUMBER'], rotation=45)

plt.xlabel('Council District Number')
plt.ylabel('Average Household Size')
plt.title('Average Household Size by District (Sorted by Size)')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```
