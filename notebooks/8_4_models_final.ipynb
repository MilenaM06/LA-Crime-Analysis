```python
import pandas as pd
dataset = pd.read_csv("../processed_data/ml_dataset.csv")
```

```python
dataset.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Features and target
X = dataset.drop(columns=['CRIME_HAPPENED'])
y = dataset['CRIME_HAPPENED']

# 80/10/10 split
X_train_val, X_test, y_train_val, y_test = train_test_split(
    X, y, test_size=0.1, random_state=42, stratify=y
)

X_train, X_val, y_train, y_val = train_test_split(
    X_train_val, y_train_val, test_size=0.1111, random_state=42, stratify=y_train_val
)

print("Shapes:")
print("X_train:", X_train.shape, "y_train:", y_train.shape)
print("X_val:", X_val.shape, "y_val:", y_val.shape)
print("X_test:", X_test.shape, "y_test:", y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression 
# Logistic Regression
lr_model = LogisticRegression(
    max_iter=1000,  # ensure convergence
    random_state=42
)

# Train
lr_model.fit(X_train, y_train)

# TEST DATASET RESULTS
y_test_pred = lr_model.predict(X_test)
print("\nLogistic Regression Test Results:")
print("Accuracy:", accuracy_score(y_test, y_test_pred))
print("Precision:", precision_score(y_test, y_test_pred))
print("Recall:", recall_score(y_test, y_test_pred))
print("F1-score:", f1_score(y_test, y_test_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier
# Random Forest Classifier
rf_model = RandomForestClassifier(
    n_estimators=300,
    max_depth=20,
    random_state=42,
    n_jobs=-1
)

# Train
rf_model.fit(X_train, y_train)

# TEST DATASET RESULTS
y_test_pred = rf_model.predict(X_test)
print("RandomForestClassifier Test Results:")
print("Accuracy:", accuracy_score(y_test, y_test_pred))
print("Precision:", precision_score(y_test, y_test_pred))
print("Recall:", recall_score(y_test, y_test_pred))
print("F1-score:", f1_score(y_test, y_test_pred))
```

```python
from catboost import CatBoostClassifier
# Cat Boost Classifier
cb_model = CatBoostClassifier(
    iterations=300,
    learning_rate=0.1,
    depth=8,
    eval_metric='F1',
    random_state=42,
    verbose=0
)

# Train
cb_model.fit(X_train, y_train, eval_set=(X_val, y_val))

# TEST DATASET RESULTS
y_test_pred = cb_model.predict(X_test)
print("CatBoostClassifier Test Results:")
print("Accuracy:", accuracy_score(y_test, y_test_pred))
print("Precision:", precision_score(y_test, y_test_pred))
print("Recall:", recall_score(y_test, y_test_pred))
print("F1-score:", f1_score(y_test, y_test_pred))
```

```python
from lightgbm import LGBMClassifier
# LightGBM Classifier
lgbm_model = LGBMClassifier(
    objective='binary',
    boosting_type='gbdt',
    learning_rate=0.1,
    num_leaves=100,
    max_depth=20,
    feature_fraction=0.8,
    bagging_fraction=0.8,
    bagging_freq=5,
    n_estimators=500,
    random_state=42,
)

# Train 
lgbm_model.fit(X_train, y_train, eval_set=(X_val, y_val))

# TEST DATASET RESULTS
y_test_pred = lgbm_model.predict(X_test)
print("LightGBM Test Results:")
print("Accuracy:", accuracy_score(y_test, y_test_pred))
print("Precision:", precision_score(y_test, y_test_pred))
print("Recall:", recall_score(y_test, y_test_pred))
print("F1-score:", f1_score(y_test, y_test_pred))
```

```python
import xgboost as xgb
# XGBoost Classifier
xgb_model = xgb.XGBClassifier(
    objective='binary:logistic',
    colsample_bytree=0.8,
    learning_rate=0.1,
    max_depth=10,
    n_estimators=300,
    subsample=1,
    eval_metric='logloss',
    random_state=42,
    n_jobs=-1
)

# Train
xgb_model.fit(X_train, y_train, eval_set=[(X_val, y_val)], verbose=False)

# TEST DATASET RESULTS
y_test_pred = xgb_model.predict(X_test)
print("XGBoost Test Results:")
print("Accuracy:", accuracy_score(y_test, y_test_pred))
print("Precision:", precision_score(y_test, y_test_pred))
print("Recall:", recall_score(y_test, y_test_pred))
print("F1-score:", f1_score(y_test, y_test_pred))
```
