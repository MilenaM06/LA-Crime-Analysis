```python
import pandas as pd
```

```python
#Crime data
crime_data = pd.read_csv("../processed_data/cleaned_crime_data.csv")
crime_data.head()
```

```python
#Weather data
weather_data = pd.read_csv("../processed_data/weather_2010_2019_hourly.csv")
weather_data.head()
```

```python
# Merge on both datetime and area
merged_data = pd.merge(
    crime_data,
    weather_data,
    left_on=['DATETIME OCC', 'AREA'],
    right_on=['datetime', 'APREC'],
    how='left'  
)

merged_data.head()
```

```python
# Drop unnecessary columns
merged_data.drop(
    columns=['AREA', 'datetime', 'APREC', 'DATETIME OCC'],
    inplace=True
)

merged_data.head()
```

```python
merged_data = merged_data.rename(columns={"temperature": "TEMPERATURE"})
merged_data = merged_data.rename(columns={"precipitation": "PERCIPITATION"})
merged_data = merged_data.rename(columns={"wind_speed": "WIND SPEED"})
```

```python
merged_data.head()
```

```python
# Export the processed DataFrame to CSV
merged_data.to_csv("../processed_data/merged_weather_crime_data.csv", index=False)
```
