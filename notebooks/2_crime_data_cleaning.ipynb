```python
# Columns to drop - they are not important
cols_to_drop = [
    "AREA ",        
    "Rpt Dist No",
    "Part 1-2",
    "Crm Cd",
    "Mocodes",
    "Premis Cd",
    "Weapon Used Cd",
    "Status",
    "Status Desc",
    "Crm Cd 1",
    "Crm Cd 2",
    "Crm Cd 3",
    "Crm Cd 4",
    "LOCATION",
    "Cross Street",
]

# Drop columns
data = data.drop(columns=[col for col in cols_to_drop if col in data.columns])

# Check the result
data.columns
```

```python
# Rename columns
data = data.rename(columns={"DR_NO": "ID"})
data = data.rename(columns={"Date Rptd": "DATE RPTD"})
data = data.rename(columns={"AREA NAME": "AREA"})
data = data.rename(columns={"Crm Cd Desc": "CRIME"})
data = data.rename(columns={"Vict Age": "VICT AGE"})
data = data.rename(columns={"Vict Sex": "VICT SEX"})
data = data.rename(columns={"Vict Descent": "VICT DESCENT"})
data = data.rename(columns={"Premis Desc": "PLACE"})
data = data.rename(columns={"Weapon Desc": "WEAPON"})
data.columns
```

```python
# Convert strings to datetime
data["DATE RPTD"] = pd.to_datetime(data["DATE RPTD"],format="%Y %b %d %I:%M:%S %p")
data["DATE OCC"] = pd.to_datetime(data["DATE OCC"], format="%Y %b %d %I:%M:%S %p")
# Format as dd-mm-yyyy
data["DATE RPTD"] = data["DATE RPTD"].dt.strftime("%Y-%m-%d")
data["DATE OCC"] = data["DATE OCC"].dt.strftime("%Y-%m-%d")

data.head()
```

```python
# Ensure time is string and pad with zeros if needed
data['TIME OCC'] = data['TIME OCC'].astype(str).str.zfill(4)
# Convert to proper HH:MM format
data['TIME OCC'] = data['TIME OCC'].str[:2] + ':' + data['TIME OCC'].str[2:]

data.head()
```

```python
# Combine DATE OCC and TIME OCC into a single datetime column
# This allows merging with weather data on an hourly basis
data.loc[:, "DATETIME OCC"] = pd.to_datetime(
    data["DATE OCC"] + " " + data["TIME OCC"],
    format="%Y-%m-%d %H:%M"
)

# Round DATETIME OCC to nearest full hour
data.loc[:, "DATETIME OCC"] = data["DATETIME OCC"].dt.floor("h")

# Drop DATE OCC and TIME OCC columns
data.drop(columns=["DATE OCC"], inplace=True)
data.drop(columns=["TIME OCC"], inplace=True)

data.head()
```

```python
def categorize_crime(crime):
    c = crime.upper()

    # 1. Violent
    if any(x in c for x in ['ASSAULT', 'ROBBERY', 'HOMICIDE', 'KIDNAPPING', 'THREAT']):
        return 'Violent'

    # 2. Sexual
    elif any(x in c for x in ['RAPE', 'SEX', 'SODOMY', 'INDECENT', 'LEWD', 'PORN', 'INCEST', 'BEAST']):
        return 'Sexual'

    # 3. Property
    elif any(x in c for x in ['BURGLARY', 'THEFT', 'SHOPLIFT', 'PICKPOCKET', 'PURSE']):
        return 'Property'

    # 4. Vehicle
    elif any(x in c for x in ['VEHICLE', 'MOTOR', 'BIKE', 'BOAT', 'DWOC']):
        return 'Vehicle'

    # 5. Fraud
    elif any(x in c for x in ['FRAUD', 'FORGERY', 'EMBEZZLEMENT', 'BUNCO', 'COUNTERFEIT']):
        return 'Fraud'

    # 6. Vandalism / Arson
    elif any(x in c for x in ['VANDALISM', 'ARSON']):
        return 'Vandalism'

    # 7. Weapons
    elif any(x in c for x in ['WEAPON', 'FIREARM', 'SHOT']):
        return 'Weapons'

    # 8. Public Order
    elif any(x in c for x in ['TRESPASS', 'DISTURB', 'PROWLER', 'RIOT', 'DISPERSE']):
        return 'Public Order'

    # 9. Family / Domestic
    elif any(x in c for x in ['CHILD', 'INTIMATE', 'DOMESTIC']):
        return 'Domestic'

    # 10. Other
    else:
        return 'Other'
```

```python
data["crime_category"] = data["CRIME"].apply(categorize_crime)
```
