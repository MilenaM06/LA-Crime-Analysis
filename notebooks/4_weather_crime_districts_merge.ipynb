```python
import pandas as pd
import geopandas as gpd
from shapely.geometry import Point
```

```python
# Define categorical data types to reduce memory usage
dtype_dict = {
    "CRIME": "category",
    "VICT SEX": "category",
    "VICT DESCENT": "category",
    "PLACE": "category",
    "WEAPON CATEGORY": "category",
}

# Load merged weather and crime dataset with optimized memory settings
weather_crime_data = pd.read_csv(
    "../processed_data/merged_weather_crime_data.csv",
    dtype=dtype_dict,
    parse_dates=["DATE RPTD", "DATETIME OCC"],
    low_memory=False
)

weather_crime_data.head()
```

```python
weather_crime_data.info()
```

```python
weather_crime = weather_crime_data.copy()
weather_crime.head()
```

```python
# Create spatial point geometry from longitude and latitude coordinates
weather_crime["GEOMETRY"] = weather_crime.apply(
    lambda row: Point(row["LON"], row["LAT"]),
    axis=1
)
```

```python
# Convert DataFrame to GeoDataFrame
weather_crime_gdf = gpd.GeoDataFrame(weather_crime, geometry="GEOMETRY", crs="EPSG:4326")
```

```python
weather_crime_gdf.head()
```

```python
# Load LA City Council district boundaries from shapefile
districts_path = "../raw_data/LA_City_Council_Districts/LA_City_Council_Districts.shp"
districts = gpd.read_file(districts_path)
```

```python
print(districts.columns)
```

```python
print(districts["DISTRICT"].unique())
```

```python
districts
```

```python
# Reproject district polygons to match crime data coordinate system
districts = districts.to_crs(weather_crime_gdf.crs)
```

```python
districts
```

```python
# Spatially join crimes with districts, assigning each crime to the district it falls within
merged_weather_crime_districts = gpd.sjoin(
    weather_crime_gdf,
    districts,
    how="left",
    predicate="within"
)

merged_weather_crime_districts.head()
```

```python
# Remove unnecessary columns
merged_weather_crime_districts = merged_weather_crime_districts.drop(columns=['GEOMETRY', 'index_right', 'OBJECTID', 'MEMBER', 'DIST_MEMBE', 'ShapeSTAre', 'ShapeSTLen'])

merged_weather_crime_districts.head()
```

```python
# Check for any invalid values in the DISTRICT column
merged_weather_crime_districts['DISTRICT'].unique()
```

```python
# Count how many crimes were not assigned to any district
merged_weather_crime_districts["DISTRICT"].isna().sum()
```

```python
# Remove rows with NaN in DISTRICT since they do not belong to any district
merged_weather_crime_districts = (
    merged_weather_crime_districts
    .dropna(subset=["DISTRICT"])
)
```

```python
# Convert DISTRICT column to integer type
merged_weather_crime_districts["DISTRICT"] = (
    merged_weather_crime_districts["DISTRICT"].astype(int)
)
```
