```python
import pandas as pd
import numpy as np
```

```python
#Positives Data Inspection
positives = pd.read_csv("../processed_data/positives.csv",  parse_dates=["DATETIME"])

positives = positives[
    [
        'DATETIME',
        'LAT',
        'LON',
        'DISTRICT',
        'TEMPERATURE',
        'PERCIPITATION',
        'WIND SPEED',
        'RAIN',
        'SNOWFALL',
        'CRIME HAPPENED',
    ]
]
positives.head()

positives.head()
```

```python
#Negatives Data Inspection
negatives = pd.read_csv("../processed_data/negatives.csv",  parse_dates=["DATETIME"])

negatives = negatives[
    [
        'DATETIME',
        'LAT',
        'LON',
        'DISTRICT',
        'TEMPERATURE',
        'PERCIPITATION',
        'WIND SPEED',
        'RAIN',
        'SNOWFALL',
        'CRIME HAPPENED',
    ]
]

negatives.head()
```

```python
# Combine datasets
dataset = pd.concat(
    [positives, negatives],
    ignore_index=True
)
```

```python
# Shuffle
dataset = dataset.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)
```

```python
# Change datetime data to be able to train model - datatime separate to columns
dataset['YEAR'] = dataset['DATETIME'].dt.year
dataset['MONTH'] = dataset['DATETIME'].dt.month
dataset['DAY'] = dataset['DATETIME'].dt.day
dataset['DAY OF WEEK'] = dataset['DATETIME'].dt.dayofweek
dataset['HOUR'] = dataset['DATETIME'].dt.hour
dataset = dataset.drop(columns=['DATETIME'])
```

```python
dataset.head()
```

```python
# Convert latitude/longitude (in degrees) to 3D Cartesian coordinates (X, Y, Z)
R = 6371 # Earth radius in km

lat_rad = np.radians(dataset['LAT'])
lon_rad = np.radians(dataset['LON'])

dataset['X'] = R * np.cos(lat_rad) * np.cos(lon_rad)
dataset['Y'] = R * np.cos(lat_rad) * np.sin(lon_rad)
dataset['Z'] = R * np.sin(lat_rad)

dataset.drop(columns=['LAT', 'LON'], inplace=True)
```

```python
dataset.head()
```

```python
from sklearn.preprocessing import OneHotEncoder

# Use one-hot encoding to convert district categories into separate binary features
encoder = OneHotEncoder(drop='first', sparse_output=False)

district_encoded = encoder.fit_transform(dataset[['DISTRICT']])

# Convert to DataFrame
district_df = pd.DataFrame(
    district_encoded,
    columns=encoder.get_feature_names_out(['DISTRICT'])
)

# Drop original column and concatenate
dataset = pd.concat([dataset.drop(columns=['DISTRICT']), district_df], axis=1)
```

```python
import numpy as np

# Encode cyclical time features (hour, month, day of week, and day of month) using sine/cosine 
# transformations to preserve their circular nature (23h is close to 0h, 31st day close to 1st)

# Hour
dataset["HOUR_sin"] = np.sin(2 * np.pi * dataset["HOUR"] / 24)
dataset["HOUR_cos"] = np.cos(2 * np.pi * dataset["HOUR"] / 24)

# Month
dataset["MONTH_sin"] = np.sin(2 * np.pi * dataset["MONTH"] / 12)
dataset["MONTH_cos"] = np.cos(2 * np.pi * dataset["MONTH"] / 12)

#Day
dataset["DAY_sin"] = np.sin(2 * np.pi * dataset["DAY"] / 31)
dataset["DAY_cos"] = np.cos(2 * np.pi * dataset["DAY"] / 31)

# Day of week
dataset["DOW_sin"] = np.sin(2 * np.pi * dataset["DAY OF WEEK"] / 7)
dataset["DOW_cos"] = np.cos(2 * np.pi * dataset["DAY OF WEEK"] / 7)

# Drop originals
dataset = dataset.drop(columns=["HOUR", "MONTH", "DAY", "DAY OF WEEK"])
```

```python
dataset.head()
```

```python
from sklearn.preprocessing import StandardScaler

# Standardize numerical features to ensure all values are on a similar scale
scaler = StandardScaler()
cols_to_scale = ["TEMPERATURE", "WIND SPEED", "RAIN", "SNOWFALL", "X", "Y", "Z"]
dataset[cols_to_scale] = scaler.fit_transform(dataset[cols_to_scale])
```
